# <span style = "color:green"> data analysis</span>

```python
df.sort_values(by='price',ascending=False)
```

```python
df.sample(10)
```

```python
df['neighbourhood_group'].value_counts().plot(kind='bar',figsize=(4,4),color='black')
plt.title("region distribution")
plt.style.use('classic')
plt.show()
```

#### #Brooklyn and Manhattan have the highest number of rented properties

```python
df['room_type'].value_counts().plot(kind='barh',figsize=(6,4),color='black')
plt.title("room type distribution",c='b')
plt.style.use('classic')
plt.show()                                    
```

#### #the most rented room types are entire home/apt and private room 

```python
sns.boxplot(data=df,x='neighbourhood_group',y='price')
plt.title("prices depend on region",c='b')
plt.style.use('classic')
plt.show()
```

#### #prices are higher in brooklyn and manhattan compared to other areas , despite being the most rented locations

```python
sns.scatterplot(data=df,x='number_of_reviews',y='price')
plt.title("effet price on reviews",c='b')
plt.style.use('dark_background')
plt.show()
```

#### #lower prices tend to have more reviews , indicating a relationship between priceand number  of reviews

```python
sns.scatterplot(data=df,x='availability_365',y='price')
plt.title("effect price on available",c='b')
plt.style.use("ggplot")
plt.show()
```

#### #lower-priced listings tend to have higher availability , suggesting a negative relationship between price and availability
